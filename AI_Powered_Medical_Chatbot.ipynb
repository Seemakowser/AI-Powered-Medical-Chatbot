```python
from transformers import pipeline
from PIL import Image
import gradio as gr

print("Libraries imported successfully!")
```

```python
classifier = pipeline(
    "image-classification",
    model="google/vit-base-patch16-224"
)

print("AI Model Loaded Successfully!")
```

```python
def explain_result(disease):

    explanations = {

        "NORMAL": """
No major heart-related abnormalities detected.

Recommendations:
• Maintain a healthy diet
• Exercise regularly
• Schedule routine health checkups
• Stay hydrated and manage stress
""",

        "HEART_DISEASE": """
 Possible signs of heart-related abnormalities detected.

Possible Concerns:
• Coronary artery disease
• Heart muscle abnormalities
• Blood circulation issues

Recommendations:
• Consult a cardiologist
• Consider ECG/ECHO examination
• Avoid smoking and excessive stress
""",

        "ARRHYTHMIA": """
 Possible irregular heartbeat pattern detected.

Recommendations:
• Monitor heart rate regularly
• Consult a healthcare professional
• Avoid excessive caffeine intake
• Follow medical advice if symptoms persist
"""
    }

    return explanations.get(
        disease,
        "Please consult a medical professional for further evaluation."
    )

print("Explanation Function Ready!")
```

```python
def predict_heart_disease(image):

    img = Image.fromarray(image)

    results = classifier(img)

    top_result = results[0]

    score = top_result['score']

    # Demo prediction logic
    if score > 0.60:
        disease = "HEART_DISEASE"
    elif score > 0.40:
        disease = "ARRHYTHMIA"
    else:
        disease = "NORMAL"

    explanation = explain_result(disease)

    output = f"""
 AI Medical Analysis Report

Prediction Result: {disease}

Confidence Score: {round(score * 100, 2)} %

{explanation}

 Disclaimer:
This project is for educational purposes only and is not a substitute for professional medical diagnosis.
"""

    return output

print("Prediction Function Ready!")
```

```python
interface = gr.Interface(
    fn=predict_heart_disease,
    inputs=gr.Image(type="numpy"),
    outputs=gr.Textbox(lines=18),
    title="AI-Powered Medical Chatbot for Heart Disease Detection",
    description="""
Upload a heart-related medical image (ECG, X-ray, MRI, etc.).
The AI system analyzes the image and provides a preliminary prediction
along with healthcare recommendations.
"""
)
```

```python
interface.launch(debug=True)
```
